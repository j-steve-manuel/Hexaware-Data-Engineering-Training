Part 2 — flatMap Practice

Create word dataset.

```python
sentences = sc.parallelize([
"big data spark",
"spark hadoop hive",
"python spark hadoop",
"spark streaming kafka"
])
```

Exercise 8\
Split all sentences into words.\
Expected concept:\

```python
sentences.flatMap(lambda x: x.split(" ")).collect()
```

Exercise 9\
Find all distinct technologies.\
Expected output example:
["spark","hadoop","hive","python","kafka"]


```python
sentences.flatMap(lambda x: x.split(" ")).distinct().collect()
```

Exercise 10\
Count number of occurrences of each word.\
Expected result format:\
spark → 4\
hadoop → 2\
python → 1\
Hint pipeline:\
flatMap\
map\
reduceByKey

```python
word_counts = sentences.flatMap(lambda x: x.split(" ")) \
    .map(lambda word: (word, 1)) \
    .reduceByKey(lambda a, b: a + b)

for word, count in word_counts.collect():
    print(f"{word} → {count}")
```

Part 6 — Advanced RDD Thinking

Exercise 29
Find department with highest total salary.

```python
totals = emp_rdd.map(lambda x: (x[1], x[2])) \
                .reduceByKey(lambda a, b: a + b)

highest_dept = totals.max(key=lambda x: x[1])

print(f"Highest Department: {highest_dept[0]} with total {highest_dept[1]}")
```

Exercise 30

Find employee count and total salary per department.

Expected output format:

IT → (count, total_salary)

```python

dept_stats = emp_rdd.map(lambda x: (x[1], (1, x[2]))) \
                    .reduceByKey(lambda a, b: (a[0] + b[0], a[1] + b[1]))

for dept, stats in dept_stats.collect():
    print(f"{dept} → {stats}")
```

Final Mini Project (RDD)

Use sales dataset.

Tasks:

1. Calculate total revenue per product

2. Find top selling product

3. Find product generating highest revenue

4. Find products sold more than 3 times

5. Sort products by revenue

Final expected output example:

Product Revenue

Laptop → 375000

Monitor → 48000

Keyboard → 7500

```python
# 1. Calculate total revenue per product
revenue_rdd = sales_rdd.map(lambda x: (x[0], x[1] * x[2])) \
                       .reduceByKey(lambda a, b: a + b)

# 2. Find top selling product (by quantity)
top_selling_product = sales_rdd.map(lambda x: (x[0], x[1])) \
                               .reduceByKey(lambda a, b: a + b) \
                               .max(key=lambda x: x[1])

# 3. Find product generating highest revenue
highest_revenue_product = revenue_rdd.max(key=lambda x: x[1])

# 4. Find products sold more than 3 times (Total quantity > 3)
frequent_products = sales_rdd.map(lambda x: (x[0], x[1])) \
                             .reduceByKey(lambda a, b: a + b) \
                             .filter(lambda x: x[1] > 3)

# 5. Sort products by revenue (Descending)
sorted_revenue = revenue_rdd.sortBy(lambda x: x[1], ascending=False)

# Final Output Display
print("Product Revenue")
for product, revenue in sorted_revenue.collect():
    print(f"{product} → {revenue}")
```

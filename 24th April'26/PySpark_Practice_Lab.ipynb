74. Rank customers by revenue.


```python
from pyspark.sql.functions import sum, col

orders.join(order_items, "order_id") \
    .join(products, "product_id") \
    .groupBy("customer_id") \
    .agg(sum(col("quantity") * col("price")).alias("total_rev")) \
    .withColumn("rank", rank().over(Window.orderBy(desc("total_rev")))).show()
```

75. Dense rank customers by revenue.


```python
from pyspark.sql.functions import dense_rank

orders.join(order_items, "order_id") \
    .join(products, "product_id") \
    .groupBy("customer_id") \
    .agg(sum(col("quantity") * col("price")).alias("total_rev")) \
    .withColumn("dense_rank", dense_rank().over(Window.orderBy(desc("total_rev")))).show()
```

Final Capstone Mini Project
Students must:
1. Join all tables
2. Calculate revenue
3. Find top customers
4. Find top products
5. Find revenue per city
6. Rank customers by revenue
7. Save final dataset



```python
from pyspark.sql import Window
from pyspark.sql.functions import col, sum, round, rank, desc

# 1. Join all tables
full_data = orders.join(order_items, "order_id") \
                  .join(products, "product_id") \
                  .join(customers, "customer_id")

# 2. Calculate revenue per line item
revenue_df = full_data.withColumn("revenue", round(col("quantity") * col("price"), 2))

# 3, 4, 5. Aggregate Analysis
top_customers = revenue_df.groupBy("customer_id", "name").agg(sum("revenue").alias("total_customer_revenue"))
top_products = revenue_df.groupBy("product_id", "product_name").agg(sum("revenue").alias("total_product_revenue"))
city_revenue = revenue_df.groupBy("city").agg(sum("revenue").alias("city_revenue"))

# 6. Rank customers by revenue
customer_window = Window.orderBy(desc("total_customer_revenue"))
ranked_customers = top_customers.withColumn("customer_rank", rank().over(customer_window))

# 7. Create the Final Report Dataset
final_df = revenue_df.join(ranked_customers, ["customer_id", "name"]) \
                     .select(
                         "order_id",
                         "name",
                         "city",
                         "product_name",
                         "revenue",
                         "total_customer_revenue",
                         "customer_rank"
                     )
```

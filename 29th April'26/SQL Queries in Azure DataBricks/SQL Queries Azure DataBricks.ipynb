```sql
-- Section 1 — Basic Queries
-- 1. Show all employees.
SELECT * FROM employees;
-- 2. Show only emp_name , city , salary .
SELECT emp_name, city, salary FROM employees;
-- 3. Show distinct departments.
SELECT DISTINCT department FROM employees;
-- 4. Count total employees.
SELECT COUNT(*) FROM employees;
-- 5. Show employees from Hyderabad.
SELECT * FROM employees WHERE city = 'Hyderabad';
-- 6. Show employees from Bangalore.
SELECT * FROM employees WHERE city = 'Bangalore';
-- 7. Show employees whose salary is greater than 90000.
SELECT * FROM employees WHERE salary > 90000;
-- 8. Show employees whose experience is greater than 6 years.
SELECT * FROM employees WHERE experience > 6;
-- 9. Show employees working in Engineering department.
SELECT * FROM employees WHERE department = 'Engineering';
-- 10. Show employees whose salary is between 80000 and 95000.
SELECT * FROM employees WHERE salary BETWEEN 80000 AND 95000;
```

```sql
-- Section 2 — Sorting
-- 11. Sort employees by salary ascending.
SELECT * FROM employees ORDER BY salary ASC;
-- 12. Sort employees by salary descending.
SELECT * FROM employees ORDER BY salary DESC;
-- 13. Show top 5 highest paid employees.
SELECT * FROM employees ORDER BY salary DESC LIMIT 5;
-- 14. Show lowest 5 salary employees.
SELECT * FROM employees ORDER BY salary ASC LIMIT 5;
-- 15. Sort employees by experience descending.
SELECT * FROM employees ORDER BY experience DESC;
```

```sql
-- Section 3 — Aggregations
-- 16. Find average salary of all employees.
SELECT AVG(salary) from employees;
-- 17. Find maximum salary.
SELECT MAX(salary) from employees;
-- 18. Find minimum salary.
SELECT MIN(salary) from employees;
-- 19. Count employees per department.
SELECT department, COUNT(*) FROM employees GROUP BY department;
-- 20. Find average salary per department.
SELECT department, AVG(salary) FROM employees GROUP BY department;
```

```sql

-- Section 4 — Group By Analytics
-- 21. Find total salary paid per department.
SELECT department, SUM(salary) FROM employees GROUP BY department;
-- 22. Find average experience per department.
SELECT department, AVG(experience) FROM employees GROUP BY department;
-- 23. Find total employees per city.
SELECT city, COUNT(*) FROM employees GROUP BY city;
-- 24. Find maximum salary per city.
SELECT city, MAX(salary) FROM employees GROUP BY city;
-- 25. Find minimum salary per department.
SELECT department, MIN(salary) FROM employees GROUP BY department;
```

```sql
-- Section 5 — Conditional Queries
-- 26. Show employees with salary greater than company average.
SELECT * FROM employees WHERE salary > (SELECT AVG(salary) FROM employees);
-- 27. Show employees with experience greater than department average.
SELECT * FROM employees e1
WHERE e1.EXPERIENCE > (SELECT AVG(EXPERIENCE) FROM employees e2 WHERE e1.department = e2.department);
-- 28. Show employees whose salary is less than 70000.
SELECT * FROM employees WHERE salary < 70000;
-- 29. Show employees working in HR department.
SELECT * FROM employees WHERE department = 'HR';
-- 30. Show employees working in Finance department with salary greater than 90000.
SELECT * FROM employees WHERE department = 'Finance' AND salary > 90000;
```

```sql
-- Section 6 — Joins
-- 31. Show employees with project budgets by joining employees and projects tables using
-- department.
SELECT e.emp_name, e.department, p.project_name, P.budget
FROM employees e JOIN projects p 
ON e.department = p.department;
-- 32. Show employee name, department, project name.
SELECT e.emp_name, e.department, p.project_name
FROM employees e JOIN projects p 
ON e.department = p.department;
-- 33. Show department-wise project budget.
SELECT e.department, SUM(p.budget) as total_budget
FROM employees e JOIN projects p 
ON e.department = p.department
GROUP BY e.department;
-- 34. Show employees working in departments that have projects.
SELECT e.emp_name, e.department
FROM employees e JOIN projects p 
ON e.department = p.department
GROUP BY e.emp_name, e.department
HAVING COUNT(p.project_id) > 0;
-- 35. Count number of employees working per project.
SELECT p.project_name, COUNT(e.emp_id) as num_employees
FROM employees e JOIN projects p 
ON e.department = p.department
GROUP BY p.project_name;



```

```sql
-- Section 7 — Advanced Queries
-- 36. Find top 3 highest paid employees in Engineering department.
SELECT * FROM employees WHERE department = 'Engineering' ORDER BY salary DESC LIMIT 3;
-- 37. Find department with highest average salary.
SELECT department, AVG(salary) as avg_salary FROM employees GROUP BY department ORDER BY avg_salary DESC LIMIT 1;
-- 38. Find city with highest number of employees.
SELECT city, COUNT(*) as num_employees FROM employees GROUP BY city ORDER BY num_employees DESC LIMIT 1;
-- 39. Find employees whose salary is higher than department average.
SELECT * FROM employees e1 
WHERE salary > (SELECT AVG(salary) FROM employees e2 
WHERE e1.department = e2.department
 GROUP BY department);
-- 40. Find employees whose experience is highest in their department.
SELECT * FROM employees E1 WHERE experience = (SELECT MAX(experience) FROM employees E2
WHERE E1.department = E2.department GROUP BY department);
```

```sql
-- Final Mini Capstone
-- Write SQL queries to create a final analytics table containing:
-- emp_name
-- city
-- department
-- salary
-- experience
-- salary_category
-- department_avg_salary

-- Rules:
-- salary_category
-- salary >= 95000 → High
-- salary >= 80000 → Medium
-- else → Low
-- Save results:

CREATE TABLE employee_salary_analysis AS
SELECT emp_name, city, department, salary, experience, 
CASE WHEN salary >= 95000 THEN 'High'
WHEN salary >= 80000 THEN 'MEDIUM'
ELSE 'Low' 
END AS salary_category,
AVG(salary) OVER (PARTITION BY department) AS department_avg_salary
FROM employees;
SELECT * FROM employee_salary_analysis;
```

```python
import pandas as pd
import numpy as np

# 1. Loading Data 

data = {
    'student_id': [1, 2, 3, 4, 5],
    'course_id': [101, 101, 102, 101, 102],
    'student_name': ['John', 'Alice', np.nan, 'Bob', 'Charlie'], # Missing name
    'enrollment_date': ['2026-05-01', '2026-05-02', '2026-05-03', None, '2026-05-05'], # Missing date
    'completion_pct': [85, 100, 45, np.nan, 120] # Missing and invalid entries (>100)
}

df = pd.DataFrame(data)

```

```python
# 2. Data Cleaning 
df = df.dropna(subset=['student_name', 'enrollment_date'])

df['completion_pct'] = df['completion_pct'].fillna(0) # Fill missing with 0
df['completion_pct'] = np.clip(df['completion_pct'], 0, 100) # Ensure range is 0-100
```

```python
# 3. NumPy Calculations 
all_progress = df['completion_pct'].values
avg_progress = np.mean(all_progress)
```

```python
# 4. Pandas Grouping 
course_summary = df.groupby('course_id')['completion_pct'].mean().reset_index()
```

```python
print("--- Cleaned Data ---")
print(df)
print(f"\nOverall Average Progress (NumPy): {avg_progress:.2f}%")
print("\n--- Course-wise Summary (Pandas) ---")
print(course_summary)
```

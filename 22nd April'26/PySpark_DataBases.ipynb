```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("PySpark Basics") \
    .getOrCreate()

print("Spark Session Created")
```

```python
data = [
    (1, "Rahul", "IT", 70000, "2024-01-10"),
    (2, "Sneha", "HR", 60000, "2024-02-15"),
    (3, "Arjun", "IT", 75000, "2024-03-01"),
    (4, "Priya", "Finance", 80000, "2024-01-25"),
    (5, "Karan", None, 50000, "2024-02-20")
]

columns = ["emp_id", "name", "department", "salary", "joining_date"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
df.show()
df.printSchema()
df.columns
df.count()
```

```python
from pyspark.sql.functions import sum, avg, max, min

df.select(
    sum("salary").alias("Total salary"),
    avg("salary").alias("Average salary"),
    max("salary").alias("Max salary"),
    min("salary").alias("Mins salary")
).show()
```

```python
dept_data = [
    ("IT", "Bangalore"),
    ("HR", "Mumbai"),
    ("Finance", "Delhi")
]

dept_columns = ["department", "location"]

dept_df = spark.createDataFrame(dept_data, dept_columns)
dept_df.show()
```

```python
data = [
    (1, "Rahul", "IT", 70000, "2024-01-10"),
    (2, "Sneha", "HR", 60000, "2024-02-15"),
    (3, "Arjun", "IT", 75000, "2024-03-01"),
    (4, "Priya", "Finance", 80000, "2024-01-25"),
    (5, "Karan", None, 50000, "2024-02-20")
]

columns = ["emp_id", "name", "department", "salary", "joining_date"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
from pyspark.sql.functions import to_date, year, month, dayofmonth

# df_dates = df.withColumn("joining_date", to_date("joining_date"))
df_dates = df.withColumns({"joining_date": to_date("joining_date")})
df_dates.show()
```

```python
df_dates.select(
    "name",
    "joining_date",
    year("joining_date").alias("year"),
    month("joining_date").alias("month"),
    dayofmonth("joining_date").alias("day")
).show()
```

UNION and UNION_BY_NAME

```python
new_data = [
    (6,"Meera","IT", 72000, "2024-04-01"),
    (7, "Amit", "HR", 50000, "2024-04-20")
 ]

new_df = spark.createDataFrame(new_data, columns)
new_df.show()
```

```python
df.union(new_df).show()
```

```python
df.unionByName(new_df).show()
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

def salary_band(Salary):
  if Salary < 50000:
    return "Low"
  elif Salary < 75000:
    return "Medium"
  else:
    return "High"

salary_band_udf = udf(salary_band, StringType())

df.withColumn("Salary Band", salary_band_udf(df.salary)).show()
```

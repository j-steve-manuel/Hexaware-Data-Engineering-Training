PART 15 — Window Functions

Exercise 44

Rank employees by salary within department.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import rank, col

windowSpec = Window.partitionBy("department").orderBy(col("salary").desc())

df_ranked = employees.withColumn("rank", rank().over(windowSpec))

df_ranked.select("department", "name", "salary", "rank").show()
```

PART 20 — Widgets in Colab

Create widgets to filter employees.

Exercise 56

Create dropdown for department.

```python
import ipywidgets as widgets
from pyspark.sql.functions import col

# Get unique department names (removing nulls)
dept_list = sorted([row['department'] for row in employees.select("department").distinct().collect() if row['department']])

# Create the dropdown widget
dept_dropdown = widgets.Dropdown(
    options=dept_list,
    value=dept_list[0],
    description='Department:',
    style={'description_width': 'initial'}
)

display(dept_dropdown)
```

Exercise 57

Create slider for salary.

```python

salary_slider = widgets.IntSlider(
    value=50000,
    min=40000,
    max=100000,
    step=5000,
    description='Min Salary:',
    style={'description_width': 'initial'},
    continuous_update=False
)

display(salary_slider)
```

Exercise 58

Display employees based on widget values.

```python
from IPython.display import display, clear_output

def filter_employees(dept, min_salary):
    print(f"--- Results for {dept} with Salary >= {min_salary} ---")

    filtered_df = employees.filter((col("department") == dept) & (col("salary") >= min_salary))

    if filtered_df.count() > 0:
        filtered_df.show()
    else:
        print("No employees match these criteria.")

widgets.interact(filter_employees, dept=dept_dropdown, min_salary=salary_slider)
```

Final Mini Project

Using all datasets, generate report:

Top Employee

Average Salary

Department Employee Count

Total Sales

Best Sales Employee

Save results to:
final_report.txt

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum, avg, count, desc


# --- Metric 1: Top Employee (Highest Salary) ---
top_emp_row = employees.orderBy(desc("salary")).first()
top_emp_name = top_emp_row["name"]
top_emp_salary = top_emp_row["salary"]

# --- Metric 2: Average Salary ---
avg_salary = employees.select(avg("salary")).collect()[0][0]

# --- Metric 3: Department Employee Count ---
# Filling null departments with "Unassigned" for the report
dept_counts = employees.fillna("Unassigned", subset=["department"]) \
                       .groupBy("department").count().collect()

# --- Metric 4: Total Sales ---
total_sales_val = sales.select(sum("amount")).collect()[0][0]

# --- Metric 5: Best Sales Employee (Highest Total Sales Amount) ---
best_sales_df = sales.groupBy("emp_id").agg(sum("amount").alias("total_sales")) \
                     .join(employees, "emp_id") \
                     .orderBy(desc("total_sales")).select("name", "total_sales").first()

best_sales_name = best_sales_df["name"]
best_sales_amount = best_sales_df["total_sales"]

# --- Generate the Report String ---
report_content = f"""--- FINAL MINI PROJECT REPORT ---
1. Top Employee (Highest Salary): {top_emp_name} (Rs. {top_emp_salary})
2. Average Salary: Rs. {avg_salary:.2f}
3. Total Sales: Rs. {total_sales_val}
4. Best Sales Employee: {best_sales_name} (Total Sales: Rs. {best_sales_amount})

5. Department-wise Employee Count:
"""

for row in dept_counts:
    report_content += f"   - {row['department']}: {row['count']}\n"

with open("final_report.txt", "w") as f:
    f.write(report_content)

print("Report generated successfully! Check final_report.txt.")
```

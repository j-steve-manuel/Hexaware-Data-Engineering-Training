Step 2 — Load the Data

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Banking Analytics").getOrCreate()

print("PySpark Session Created")
```

```python
customers = spark.read.csv("bank_customers.csv", header=True, inferSchema = True)
accounts = spark.read.csv("accounts.csv", header=True, inferSchema=True)
transactions = spark.read.csv("transactions.csv", header=True, inferSchema = True)
branches = spark.read.csv("branches.csv", header=True, inferSchema=True)
profiles = spark.read.json("customer_profiles.json", multiLine=True)
```

Section 10 — Window Functions

86. Rank customers by balance within city.


```python
from pyspark.sql import Window
from pyspark.sql.functions import col, rank, row_number, dense_rank, desc

from pyspark.sql.functions import sum
```

```python
cust_acc = customers.join(accounts, "customer_id")

window_city = Window.partitionBy("city").orderBy(desc("balance"))
cust_acc.withColumn("rank", rank().over(window_city)).show()
```

87. Use row_number to find top balance account per city.


```python
cust_acc.withColumn("rn", row_number().over(window_city)) \
    .filter(col("rn") == 1).show()
```

88. Use dense_rank to rank transaction amount by transaction type.


```python
window_txn_type = Window.partitionBy("txn_type").orderBy(desc("amount"))
transactions.withColumn("d_rank", dense_rank().over(window_txn_type)).show()
```

89. Find top 2 transactions per account.


```python
window_acc = Window.partitionBy("account_id").orderBy(desc("amount"))

transactions.withColumn("rn", row_number().over(window_acc)) \
    .filter(col("rn") <= 2).show()

```

90. Create running transaction total per account_id.


```python
window_run = Window.partitionBy("account_id").orderBy("txn_date")
transactions.withColumn("running_total", sum("amount").over(window_run)).show()
```

91. Rank branches by total balance.


```python
branch_totals = accounts.groupBy("branch").agg(sum("balance").alias("total_bal"))
branch_totals.withColumn("rank", rank().over(Window.orderBy(desc("total_bal")))).show()
```

92. Rank cities by total customer balance.


```python

city_bal = cust_acc.groupBy("city").agg(sum("balance").alias("city_total"))
city_bal.withColumn("rank", rank().over(Window.orderBy(desc("city_total")))).show()
```

93. Rank customers by total transaction amount.


```python
cust_txns = cust_acc.join(transactions, "account_id")
cust_total_txn = cust_txns.groupBy("name").agg(sum("amount").alias("grand_total"))
cust_total_txn.withColumn("rank", rank().over(Window.orderBy(desc("grand_total")))).show()
```

94. Find highest transaction customer per city.


```python
full_df = cust_acc.join(transactions, "account_id")
window_city_txn = Window.partitionBy("city").orderBy(desc("amount"))

full_df.withColumn("rn", row_number().over(window_city_txn)) \
    .filter(col("rn") == 1) \
    .select("city", "name", "amount").show()
```

95. Find highest balance customer per region.

```python
reg_df = cust_acc.join(branches, "branch")
window_reg = Window.partitionBy("region").orderBy(desc("balance"))

reg_df.withColumn("rn", row_number().over(window_reg)) \
    .filter(col("rn") == 1) \
    .select("region", "name", "balance").show()
```

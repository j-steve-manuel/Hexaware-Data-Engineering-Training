80. Explain why MERGE is used for incremental loads.

The MERGE operation is the gold standard for data engineering because it solves the "Upsert" problem:

Atomicity: The update and insert happen in one transaction; if it fails, the data isn't corrupted.

Performance: It only modifies the files containing changed records, rather than rewriting the entire multi-terabyte table.

Data Integrity: It prevents the creation of duplicate records for the same ID, ensuring your "Source of Truth" remains accurate.

87. Try to explain why update is easier on Delta than raw Parquet.


Updating raw Parquet is difficult because it is an immutable format.

To change just one row in a Parquet file, we have to:

*  Read the entire file into memory.
*  Change the value in the DataFrame.
*  Delete the old file on disk.
*  Write a new file with the change.


Delta Lake simplifies this via "Copy-on-Write" or "Merge-on-Read":

1. Transaction Log: Delta uses a log to keep track of which files are valid.

2. File Management: When you update a row, Delta writes a new version of only the file that contained that row and marks the old file as "tombstoned" (inactive) in the log.

3. Consistency: Readers never see a partial update; they only see the table as it existed at a specific transaction ID.

```python
import dlt
from pyspark.sql.functions import *

@dlt.table(
    name="bronze_patient_visits",
    comment="Raw ingestion of visit and payment data"
)
def bronze_patient_visits():
    # In a real scenario, this would be spark.readStream
    # For this assessment, we create a union of the base datasets
    return visits_df.join(payments_df, "visit_id", "inner")
```

89. Create Silver table with cleaned fields.


```python
@dlt.table(
    name="silver_patient_visits",
    comment="Cleaned and transformed visit records"
)
@dlt.expect_or_drop("valid_bill", "bill_amount > 0")
@dlt.expect_or_drop("valid_status", "visit_status != 'Cancelled'")
```

90. Add total bill calculation in Silver.


91. Remove invalid records in Silver.


```python
def silver_patient_visits():
    return dlt.read("bronze_patient_visits") \
        .withColumn("visit_date", col("visit_date").cast("date")) \
        .withColumn("total_bill_with_tax", col("bill_amount") * 1.18) \
        .select(
            "visit_id", "patient_id", "doctor_id",
            "visit_date", "visit_status", "bill_amount", "total_bill_with_tax"
        )
```

92. Create Gold table grouped by city.


```python
@dlt.table(
    name="gold_city_revenue",
    comment="Aggregate revenue summarized by city"
)
def gold_city_revenue():
    return dlt.read("silver_patient_visits") \
        .join(patients_df, "patient_id") \
        .groupBy("city") \
        .agg(sum("bill_amount").alias("total_city_revenue"))
```

93. Create Gold table grouped by specialization.


```python
@dlt.table(
    name="gold_specialization_revenue",
    comment="Aggregate revenue summarized by doctor specialization"
)
def gold_specialization_revenue():
    return dlt.read("silver_patient_visits") \
        .join(doctors_df, "doctor_id") \
        .groupBy("specialization") \
        .agg(sum("bill_amount").alias("total_spec_revenue"))
```


80. Explain why MERGE is important.

The MERGE operation is critical for maintaining a Single Source of Truth in a production environment.

**Idempotency:** You can run the same merge multiple times without creating duplicate records; it simply ensures the target matches the source.

**Efficiency:** Instead of overwriting the entire table (which is slow and resource-heavy), Delta only modifies the specific files containing the affected rows.

**Data Integrity:** It handles complex logic—like SCD (Slowly Changing Dimensions)—in a single transaction. If the merge fails halfway through, Delta rolls back the change so your data never stays in a "broken" state.

**Simplification:** It removes the need for complex "Check if exists, then update, else insert" code blocks, condensing it into one readable SQL or PySpark command.

85. Compare Parquet vs Delta behavior.


The biggest difference is Transactionality.

Parquet: If you try to update a single row, you technically have to read the whole file, change the row in memory, and overwrite the entire file.

Delta: You can use UPDATE, DELETE, and MERGE commands directly because Delta manages the file versions for you.

87. Explain why Delta supports updates better.

Delta Lake is superior for updates because of the Transaction Log (_delta_log).

**ACID Compliance:** When you update a row, Delta doesn't "change" the old Parquet file. It writes a new file with the updated data and adds a JSON entry to the log saying, "As of Version 2, ignore the old file and use this new one."

**Atomicity:** If the update fails, the log entry is never written, so the table stays in its previous healthy state. Parquet, by contrast, could end up with corrupted or partial data if a write is interrupted.

**Schema Evolution:** Delta allows you to update the table structure (adding columns) without breaking downstream processes, which is very difficult to manage with raw Parquet files.

```python
import pandas as pd
import numpy as np
import json

# 1. Load Data

json_data = '''
[
  {"product_id": 101, "quantity_adjusted": -2, "timestamp": "2026-05-14T10:30:00Z", "reorder_level": 10},
  {"product_id": 101, "quantity_adjusted": 50, "timestamp": "2026-05-12T09:00:00Z", "reorder_level": 10},
  {"product_id": 205, "quantity_adjusted": 5, "timestamp": "2026-05-14T11:45:00Z", "reorder_level": 20},
  {"product_id": 205, "quantity_adjusted": -3, "timestamp": "2026-05-13T08:15:00Z", "reorder_level": 20}
]
'''
df = pd.DataFrame(json.loads(json_data))
```

```python
# 2. Data Cleaning & Transformation

df['timestamp'] = pd.to_datetime(df['timestamp'])

# Quantity Validation
df['quantity_adjusted'] = df['quantity_adjusted'].astype(np.int32)
```

```python
# 3. Calculate Current Stock Levels using NumPy

stock_summary = df.groupby('product_id').agg({
    'quantity_adjusted': 'sum',
    'reorder_level': 'first' 
}).rename(columns={'quantity_adjusted': 'current_stock'})
```

```python
# 4. Identify Low Stock Items

low_stock_items = stock_summary[stock_summary['current_stock'] < stock_summary['reorder_level']]

print("--- Inventory Summary ---")
print(stock_summary)

print("\n--- Items Requiring Reorder ---")
print(low_stock_items)
```
